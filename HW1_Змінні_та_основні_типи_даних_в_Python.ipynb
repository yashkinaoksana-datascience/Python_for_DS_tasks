**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
var1 = 15
var2 = 4
```

```python
sum = var1 + var2
difference = var1 - var2
prod= var1 * var2
div = var1 / var2
integer_div = var1 // var2
remainder = var1 / var2 - integer_div
exponentiation = var1 ** var2
```

```python
print(f"Сума {var1} та {var2} дорівнює {sum}")
print(f"Різниця {var1} та {var2} дорівнює {difference}")
print(f"Добуток {var1} та {var2} дорівнює {prod}")
print(f"Ділення {var1} на {var2} дорівнює {div}")
print(f"Цілочисельне ділення {var1} на {var2} дорівнює {integer_div}")
print(f"Залишок від ділення {var1} на {var2} дорівнює {remainder}")
print(f"Піднесення {var1} до степеня {var2} дорівнює {exponentiation}")
```
