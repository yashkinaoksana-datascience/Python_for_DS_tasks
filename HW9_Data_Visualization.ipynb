**1. Гістограма розподілу рахунків**
Завантажте датасет Tips (Чайові) з файла у матеріалах `tips.csv`.
Створіть гістограму для колонки `total_bill` з 20 bins.
Опціонально можна додати підписи осей ОХ, ОУ, а також назву графіку.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('tips.csv')
```

```python
# Колонки total_bill з 20 bins. Опціонально додано підписи осей ОХ, ОУ, а також назву графіку

df['total_bill'].hist(bins =20, edgecolor = 'black')
plt.title('Розподіл загальної суми рахунку')
plt.xlabel('Сума рахунку')
plt.ylabel('Кількість')
plt.show()
```

---

**2. Box plot чайових за днями тижня**

Створіть box plot, який показує розподіл чайових (`tip`) за днями тижня (`day`).
В який день навище медіанне значення чайоивх? Зелена горизонтальна лінія всередині кожного box-plot.


```python
df.boxplot(column = 'tip', by = 'day', figsize = (10, 6))
plt.title('Розподіл чайових за днями')
plt.suptitle('')
plt.ylabel('Чайові')
plt.show()
```

```python
df.groupby('day')['tip'].median()
```

**3. Scatter plot: рахунок vs чайові**

Створіть діаграму розсіювання для `total_bill` та `tip`.

```python
df.plot.scatter(x = 'total_bill', y = 'tip', alpha = 0.5)
plt.title('Залежність між загальною сумою рахунку та розміром чайових')
plt.xlabel('Загальна сума замовлення')
plt.ylabel('Розмір чайових')
plt.show()
```

---

**4. Bar chart: середній рахунок за часом**

Створіть стовпчасту діаграму середнього рахунку за часом дня (`time`).

```python
# Стовпчаста діаграму середнього рахунку за часом відвідування (time)

print(df.groupby('time')['total_bill'].mean())

df.groupby('time')['total_bill'].mean().plot(kind = 'bar', color = 'skyblue', edgecolor = 'black')
plt.title('Середній рахунок за часом відвідування')
plt.xlabel('Час відвідування')
plt.ylabel('Середній рахунок')
plt.xticks(rotation = 0)
plt.show()
```

**5. Інтерактивна діаграма розсіювання**

Створіть з допомогою бібліотек Plotly інтерактивний scatter plot для `total_bill` vs `tip` з кольором за статтю (`sex`).


```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
# Інтерактивний scatter plot для total_bill vs tip з кольором за статтю (sex).

fig = px.scatter(df, x = 'total_bill', y = 'tip', color = 'sex',
                 title = 'Залежність між загальною сумою рахунку та розміром чайових',
                 labels = {'total_bill': 'Загальна сума рахунку',
                          'tip': 'Розмір чайових',
                          'sex': 'Стать'},
                 color_discrete_sequence=['#1f77b4', '#ff7f0e'])

fig.show()
```

---

**6. Інтерактивний box plot: чайові за днями та статтю**

Створіть інтерактивний box plot для чайових за днями тижня з розбивкою за кольором по статі.

```python
# Інтерактивний box plot для чайових за днями тижня з розбивкою за кольором по статі

fig = px.box(df, x = 'day', y = 'tip', color = 'sex',
                 title = 'Залежність днем тижня та розміром чайових',
                 labels = {'day': 'День тижня',
                          'tip': 'Розмір чайових',
                          'sex': 'Стать'},
                 color_discrete_sequence=['#1f77b4', '#ff7f0e'])

fig.show()
```
