**2. Базові операції**

Створіть Numpy-масив зі значеннями [12, 24, 36, 48, 60].

Виконайте наступні операції над **оригінальним** масивом і запишіть у нову змінну результат (кожен пункт оцінюється в 0.5 бала):
- Помножте всі елементи на 2
- Додайте до всіх елементів 10
- Піднесіть всі елементи до квадрату
- Обчисліть квадратний корінь з усіх елементів

Виведіть результат виконання кожної операції.

```python
arr_new = np.array([12, 24, 36, 48, 60])
arr_new
```

```python
arr_new_dot_2 = arr_new * 2
arr_new_ad_10 = arr_new + 10
arr_new_squared = arr_new ** 2
arr_new_sqr = np.sqrt(arr_new)

print(f'Всі елементи помножені на 2 {arr_new_dot_2}')
print(f'До всіх елементів додано 10 {arr_new_ad_10}')
print(f'Всі елементи піднесено в квадрат {arr_new_squared}')
print(f'Квадратний корінь з елементів масива {arr_new_sqr}')
```

**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

Обчисліть використовуючи функціонал numpy (кожен пункт оцінюється в 0.5 бала):
- Середнє значення
- Мінімум
- Максимум
- Суму всіх елементів

```python
data = np.array([18, 27, 33, 41, 52, 68, 79])
data
```

```python
print(f'Середнє значення {np.mean(data):.2f}')
print(f'Мінімум {np.min(data)}')
print(f'Максимум {np.max(data)}')
print(f'Сума всіх елементів {np.sum(data)}')
```

**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3
- Елементи більші за 40

```python
arr = np.array([15, 22, 35, 44, 58, 67])
```

```python
print(f'Перший елемент {arr[0]}')
print(f'Останній елемент {arr[-1]}')
print(f'Елементи з індексами 1 до 3 {arr[1:4]}')
mask = arr > 40
print(f'Елементи більші за 40 {arr[mask]}')
```

**5. Двовимірні масиви**

Задано матрицю 3x3.

Виконайте наступні дії з функціоналом Numpy (кожен пункт оцінюється в 0.5 бала):
- Виведіть форму (shape) матриці
- Виведіть перший рядок
- Виведіть другу колонку
- Обчисліть суму всіх елементів
- Обчисліть середнє всіх елементів

```python
matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])
```

```python
print(f'Форма матриці {matrix.shape}')
print(f'Перший рядок {matrix[0, :]}') # Виводимо перший, який бачимо, а не перший в нумерації рядків
print(f'Другий стовпчик {matrix[:, 1]}') # Виводимо другий стовпчик, який бачимо, а не другий в нумерації, тобто останній, який бачимо
print(f'Сума всіх елементів {np.sum(matrix)}')
print(f'Середнє {np.mean(matrix)}')
```
